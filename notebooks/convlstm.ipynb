# ConvLSTM demo (benchmark B2)

Trains the reported architecture (`dilated_seq3`: sequence length 3, 32 hidden channels, 5x5 kernel, dilation 4, one recurrent layer) for a couple of epochs and evaluates it. This is a smoke test of the pipeline, not the reported accuracy.

For the full training budget and every other variant this driver runs, see the main [README's Training section](../README.md#training-a-model).

Requires `GW_DATA` set before launching Jupyter, or the data placed under `data/` at the repository root.

```python
import os
import subprocess
from pathlib import Path

REPO = Path.cwd().parent  # this notebook lives in notebooks/
DATA = Path(os.environ.get("GW_DATA", REPO / "data"))
DRIVER = (REPO / "experiments" / "architecture_study" /
          "run_convlstm_architecture_study.py")

print("repository:", REPO)
print("data root: ", DATA)
```

```python
subprocess.run(
    ["python", str(DRIVER),
     "--benchmark", "b2", "--arm", "unconstrained",
     "--variant", "dilated_seq3", "--epochs", "2"],
    check=True,
)
```

## Conservation diagnostics on the test predictions

```python
PRED_DIR = (REPO / "experiments" / "architecture_study" / "outputs" /
            "b2" / "dilated_seq3_h32_k5_d4_l1_rf23-55" /
            "unconstrained" / "seed42" / "one_step_predictions")

subprocess.run(
    ["python", str(REPO / "src" / "diagnostics" / "evaluate_b2.py"),
     "--prediction-dir", str(PRED_DIR),
     "--output-dir", str(REPO / "out" / "convlstm_demo"),
     "--model-name", "ConvLSTM B2 demo",
     "--prediction-mode", "one-step", "--no-plots"],
    check=True,
)
```

The reported ConvLSTM reaches R² > 0.9996 with a residual ratio near 289 and a mass-balance error near 114% of the well rate -- see the main README's results table.
